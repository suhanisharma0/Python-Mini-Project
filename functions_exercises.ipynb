```python
#Exercise 2: Shopping Discount Calculator
#Create a function named calculate_discount() that accepts a bill amount and an optional discount percentage. Apply these conditions:

#Bill amount ≤ ₹0 → Display "Invalid Bill Amount".
#Bill < ₹1,000 → Apply 5% discount.
#Bill from ₹1,000 to ₹4,999 → Apply 10% discount.
#Bill from ₹5,000 to ₹9,999 → Apply 15% discount.
#Bill ≥ ₹10,000 → Apply 20% discount.
#If the customer provides a custom discount, use it instead of the default discount.
#The custom discount must be between 0% and 100%.
#Display the original bill, discount percentage, discount amount, and final payable amount.
```

```python
def calculate_discount(bill_amount, discount=None):

    if bill_amount <= 0:
        print("Invalid Bill Amount")
        return

    if discount is not None:
        if discount < 0 or discount > 100:
            print("Invalid Discount Percentage")
            return
    else:
        if bill_amount < 1000:
            discount = 5
        elif bill_amount < 5000:
            discount = 10
        elif bill_amount < 10000:
            discount = 15
        else:
            discount = 20

    discount_amount = bill_amount * discount / 100
    final_amount = bill_amount - discount_amount

    print("Original Bill:", bill_amount)
    print("Discount Percentage:", discount, "%")
    print("Discount Amount:", discount_amount)
    print("Final Payable Amount:", final_amount)


calculate_discount(800)
print()

calculate_discount(3000)
print()

calculate_discount(12000)
print()

calculate_discount(2000, 25)
```

```python
#Exercise 3: Employee Details Using *args and **kwargs
#Create a function that accepts any number of employee details, such as department, salary, city, and experience.
```

```python
def employee_details(*args, **kwargs):
    print("Employee Details:")

    for item in args:
        print(item)

    for key, value in kwargs.items():
        print(key, ":", value)


employee_details(
    "Employee 101",
    "Full Time",
    department="IT",
    salary=60000,
    city="Bareilly",
    experience=2
)
```

```python
#Exercise 4: Student Result Function
#Create a student result function that accepts the student's name, any number of subject marks, and additional information such as roll number and semester.
```

```python
def student_result(name, *marks, **details):
    print("Student Name:", name)
    print("Subject Marks:", marks)

    if len(marks) == 0:
        print("No marks provided")
        return

    total = sum(marks)
    average = total / len(marks)

    print("Total Marks:", total)
    print("Average Marks:", average)

    for key, value in details.items():
        print(key, ":", value)


student_result(
    "Niharika",
    85, 90, 78, 88,
    roll_no=101,
    semester=1
)
```

```python
#Exercise 5: Shopping Application Using Argument Unpacking
#A shopping application stores prices in a list and customer details in a dictionary. Pass them to functions using argument unpacking.
```

```python
def calculate_total(*prices):
    total = sum(prices)
    print("Total Bill:", total)


def customer_details(**details):
    print("Customer Details:")

    for key, value in details.items():
        print(key, ":", value)


prices = [500, 1000, 1500]

customer = {
    "name": "Kushagra",
    "city": "Dehradun",
    "membership": "Gold"
}

calculate_total(*prices)

customer_details(**customer)
```

```python
#Exercise 6: Calculate Average Marks
#A training institute needs a function that calculates the average marks of a student. Reject invalid marks and empty input.

#For this example, assume valid marks are between 0 and 100, inclusive.
```

```python
def calculate_average(*marks):

    if len(marks) == 0:
        print("Error: No marks provided")
        return

    for mark in marks:
        if mark < 0 or mark > 100:
            print("Error: Invalid marks")
            return

    average = sum(marks) / len(marks)

    print("Marks:", marks)
    print("Average Marks:", average)


calculate_average(80, 90, 70, 85)

calculate_average(80, 120, 70)

calculate_average()
```
